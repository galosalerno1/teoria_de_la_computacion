**¿Por qué se dice que una MTU es capaz de simular cualquier Máquina de Turing?**

```python
Una Máquina de Turing Universal (MTU) es una máquina que puede simular cualquier otra Máquina de Turing. Recibe como entrada la descripción codificada de una máquina M (⟨M⟩) y una cadena w, y simula paso a paso el comportamiento de M sobre w, dando el mismo resultado que M dejaría en su cinta: U(⟨M, w⟩) = M(w). Es decir, U acepta si M acepta w y rechaza si M rechaza w.

Esto es posible porque los estados, símbolos y transiciones de M se pueden codificar (por ejemplo, en binario) como datos que U puede leer. Así, U no está diseñada para un problema específico: M actúa como un programa y U como quien lo ejecuta. Por eso se la llama universal o de propósito general.
```

## 1. Codificación de una máquina simple

**Máquina elegida: M inversora de bits.** Recorre la cinta de izquierda a derecha reemplazando cada 0 por 1 y cada 1 por 0. 
Se detiene al llegar al final de la cadena, porque no hay transición definida para el blanco.

* Alfabeto: {0, 1}
* Estados: {q0}
* Estado inicial: q0
* Estados finales: {} (como en el ejemplo visto en clase, la máquina termina cuando no hay transición)

| Q  | 0        | 1        |
|----|----------|----------|
| q0 | (q0,1,R) | (q0,0,R) |

### Codificación numérica

* Símbolos: `0 = 0`, `1 = 1`
* Estados: `q0 = 00`
* Movimientos: `L = 1`, `R = 0`

Cada transición se codifica como *estado nuevo + símbolo a escribir + movimiento*:

| Q  | 0    | 1    |
|----|------|------|
| 00 | 0010 | 0000 |

* $\delta(q_0, 0) = (q_0, 1, R)$ → `(00, 0, 00, 1, 0)` → registro `0000010`
* $\delta(q_0, 1) = (q_0, 0, R)$ → `(00, 1, 00, 0, 0)` → registro `0010000`

### Entrada para la MTU

Para $w$ = `1011`, con la cabeza lectora/grabadora sobre el primer símbolo:

`*011$001#0000010#0010000`

### Recorrido paso a paso de la MTU (w = 1011)

La MTU parte de la cinta `*011$001#0000010#0010000` y, en cada paso:

1. Toma el estado y el símbolo que lee la cabeza lectora/grabadora (lo que está entre `$` y el primer `#`).
2. Busca el registro que empieza con esos 3 dígitos.
3. Escribe el símbolo indicado en el lugar del `*`, mueve la cabeza y actualiza el estado y el símbolo leído.

| Paso | Cinta codificada | Qué pasó |
|------|------------------|----------|
| Inicio | `*011$001#0000010#0010000` | q0 (`00`) lee 1 |
| 1 | `0*11$000#0000010#0010000` | Registro `0010000`: escribe 0 y mueve a la derecha. Ahora lee 0 |
| 2 | `01*1$001#0000010#0010000` | Registro `0000010`: escribe 1 y mueve a la derecha. Ahora lee 1 |
| 3 | `010*$001#0000010#0010000` | Registro `0010000`: escribe 0 y mueve a la derecha. Ahora lee 1 |
| 4 | `0100*$00#0000010#0010000` | Registro `0010000`: escribe 0 y mueve a la derecha. La cabeza queda sobre el blanco |
| Fin | - | No existe registro para q0 con blanco, por lo que la MTU se detiene |

**Resultado:** la cinta final de M es `0100`, es decir, `1011` con todos sus bits invertidos.


2. **Simulación básica**  
   * Implementar en Python un programa que reciba:
     * La codificación de una máquina $M$
     * Una cadena de entrada $w$
   * El programa debe simular paso a paso la ejecución de $M$ sobre $w$

```python
def parsear_entrada(entrada_codificada):
    """Separa la entrada de la MTU en: cinta de M, posicion de la cabeza,
    estado actual y lista de registros (transiciones codificadas)."""
    izquierda, resto = entrada_codificada.split('$', 1)
    partes = resto.split('#')
    estado_simbolo = partes[0]
    registros = partes[1:]

    estado_actual = estado_simbolo[0:2]      # 2 bits de estado
    simbolo_tapado = estado_simbolo[2]       # 1 bit de simbolo

    cinta = list(izquierda)
    pos_cabeza = cinta.index('*')
    cinta[pos_cabeza] = simbolo_tapado       # se restituye el simbolo que tapaba el '*'

    return cinta, pos_cabeza, estado_actual, registros


def parsear_registros(registros):
    """Convierte cada registro de 7 bits en una transicion:
    (estado, simbolo_leido) -> (estado_nuevo, simbolo_escribe, movimiento)"""
    transiciones = {}
    for r in registros:
        estado = r[0:2]
        simbolo = r[2]
        nuevo_estado = r[3:5]
        escribe = r[5]
        movimiento = r[6]
        transiciones[(estado, simbolo)] = (nuevo_estado, escribe, movimiento)
    return transiciones


def mostrar_cinta(cinta, pos_cabeza):
    """Devuelve la cinta como texto, marcando la posicion de la cabeza con []"""
    vista = cinta.copy()
    simbolo_actual = vista[pos_cabeza] if pos_cabeza < len(vista) else '_'
    marcada = ''.join(vista[:pos_cabeza]) + '[' + simbolo_actual + ']' + ''.join(vista[pos_cabeza+1:])
    return marcada


def simular(entrada_codificada, max_pasos=200, verbose=True):
    """Simula paso a paso una MTU a partir de una entrada codificada
    con el formato *, $, # (cinta w a la izquierda, M a la derecha)."""
    cinta, pos_cabeza, estado, registros = parsear_entrada(entrada_codificada)
    transiciones = parsear_registros(registros)

    paso = 0
    while paso < max_pasos:
        simbolo = cinta[pos_cabeza] if pos_cabeza < len(cinta) else '_'

        if verbose:
            print(f"Paso {paso}: estado={estado}  cinta={mostrar_cinta(cinta, pos_cabeza)}")

        clave = (estado, simbolo)
        if clave not in transiciones:
            if verbose:
                print(f"No hay transicion para (estado={estado}, simbolo={simbolo}) -> la MTU se detiene")
            break

        nuevo_estado, escribe, movimiento = transiciones[clave]

        if pos_cabeza < len(cinta):
            cinta[pos_cabeza] = escribe
        else:
            cinta.append(escribe)

        estado = nuevo_estado
        pos_cabeza += 1 if movimiento == '0' else -1
        paso += 1

    resultado = ''.join(cinta)
    return resultado, estado
def construir_entrada(w, pos_cabeza, estado_inicial_bin, registros):
    """Arma la entrada codificada de la MTU a partir de w, reemplazando
    el simbolo en pos_cabeza por '*' y guardandolo junto al estado inicial."""
    simbolo_tapado = w[pos_cabeza]
    izquierda = w[:pos_cabeza] + '*' + w[pos_cabeza+1:]
    return f"{izquierda}${estado_inicial_bin}{simbolo_tapado}#{registros}"

# Prueba con la maquina inversora de bits, w ingresada por teclado
w = input("Ingresa la cadena binaria w (por ejemplo 1100): ")
entrada = construir_entrada(w, 0, "00", "0000010#0010000")
resultado, estado_final = simular(entrada)
print(f"\nResultado final: {resultado}  (estado final: {estado_final})")
```

En el TP utilice una codificación binaria: la cinta de la MTU se divide en dos partes con el símbolo $. A la izquierda va la cinta de M, con un * marcando dónde está la cabeza. A la derecha va el estado actual más el símbolo que tapó el *, y después los registros de transiciones separados por #. Cada registro tiene 7 bits: 2 de estado actual, 1 de símbolo leído, 2 de estado nuevo, 1 de símbolo a escribir y 1 de movimiento (R=0, L=1).

La MTU demuestra que un programa puede tratarse como un dato más. En vez de tener una máquina distinta para cada tarea, hay una sola máquina (U) que lee la descripción de cualquier otra (M) y hace lo mismo que haría M. Eso es exactamente lo que hace una computadora: la CPU no sabe de antemano qué programa va a correr, simplemente lee instrucciones guardadas en memoria y las ejecuta. Esa idea se llama programa almacenado, y es la base de las computadoras desde la arquitectura de von Neumann en adelante."
